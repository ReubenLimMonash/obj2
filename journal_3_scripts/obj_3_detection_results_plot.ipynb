```python
# For gamma

FN_INFO_FILE = "/home/research-student/reuben_ws/omnet-fanet/results_new_Mar25/gamma_failure_detection_fn_999.csv"
DETECTION_RESULT_FILE = "/home/research-student/reuben_ws/omnet-fanet/results_new_Mar25/gamma_failure_detection_runs_results.csv"
POS_LABEL_TH = [0, 26, 53]
REL_TH = 0.999

'''For convergence (L=100)'''
# For runs within a range
START_RUN_NUM = 0
END_RUN_NUM = 499

fn_df = pd.read_csv(FN_INFO_FILE)
detection_df = pd.read_csv(DETECTION_RESULT_FILE)
detection_df = detection_df.loc[~(detection_df["Sensitivity"].isnull())] # If sensitivity/mFNR is null, means no false negative

fn_groups = fn_df.groupby(["Scenario", "Combination"])
detection_groups = detection_df.groupby(["Scenario", "Combination"])
gamma_fn_group_mean_rel = [[], [], []] # To store the mean measured reliability of all links below REL_TH in group (combination)
for i in range(len(POS_LABEL_TH)):
    for group in tqdm(set(fn_groups.groups.keys())):
        fn_group = fn_groups.get_group(group)
        detection_group = detection_groups.get_group(group)
        # For convergence 
        fn_group = fn_group.loc[(fn_group["Run_Num"]>=START_RUN_NUM) & (fn_group["Run_Num"]<=END_RUN_NUM)]
        # Group by runs
        fn_runs = fn_group.groupby(["Run_Num"])
        detection_runs = detection_group.groupby(["Run_Num"])
        for run_num in set(fn_runs.groups.keys()):
            fn_run = fn_runs.get_group(run_num)
            detection_run = detection_runs.get_group(run_num)
            if detection_run["Num_Pos_Label"].values[0] > POS_LABEL_TH[i]:
                # Get the link reliabilities that are below the required level
                ul_rel = fn_group["UL_Measured_Reliability_1"].to_numpy()
                ul_fn_rel = ul_rel[np.where(ul_rel < REL_TH)]
                uav_0_rel = fn_group["UAV_0_Measured_Reliability_1"].to_numpy()
                uav_0_fn_rel = uav_0_rel[np.where(uav_0_rel < REL_TH)]
                uav_1_rel = fn_group["UAV_1_Measured_Reliability_1"].to_numpy()
                uav_1_fn_rel = uav_1_rel[np.where(uav_1_rel < REL_TH)]
                uav_2_rel = fn_group["UAV_2_Measured_Reliability_1"].to_numpy()
                uav_2_fn_rel = uav_2_rel[np.where(uav_2_rel < REL_TH)]
                uav_3_rel = fn_group["UAV_3_Measured_Reliability_1"].to_numpy()
                uav_3_fn_rel = uav_3_rel[np.where(uav_3_rel < REL_TH)]
                uav_4_rel = fn_group["UAV_4_Measured_Reliability_1"].to_numpy()
                uav_4_fn_rel = uav_4_rel[np.where(uav_4_rel < REL_TH)]
                uav_5_rel = fn_group["UAV_5_Measured_Reliability_1"].to_numpy()
                uav_5_fn_rel = uav_5_rel[np.where(uav_5_rel < REL_TH)]
                uav_6_rel = fn_group["UAV_6_Measured_Reliability_1"].to_numpy()
                uav_6_fn_rel = uav_6_rel[np.where(uav_6_rel < REL_TH)]
                uav_7_rel = fn_group["UAV_7_Measured_Reliability_1"].to_numpy()
                uav_7_fn_rel = uav_7_rel[np.where(uav_7_rel < REL_TH)]

                gamma_fn_group_mean_rel[i].append(np.nanmean(np.hstack((ul_fn_rel, uav_0_fn_rel, uav_1_fn_rel, uav_2_fn_rel, uav_3_fn_rel,
                                                            uav_4_fn_rel, uav_5_fn_rel, uav_6_fn_rel, uav_7_fn_rel))))
```
