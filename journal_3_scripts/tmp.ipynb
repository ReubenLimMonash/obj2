```python
import os
import pandas as pd
from multiprocessing.pool import Pool
from itertools import repeat

filenames = ["data_manual_throughput_train_new_UAV_1_processed.csv", "data_manual_throughput_train_new_UAV_2_processed.csv",
             "data_manual_throughput_train_new_UAV_3_processed.csv", "data_manual_throughput_train_new_UAV_4_processed.csv",
             "data_manual_throughput_train_new_UAV_5_processed.csv", "data_manual_throughput_train_new_UAV_6_processed.csv",
             "data_manual_throughput_train_new_UAV_7_processed.csv", "data_manual_throughput_train_new_ul_processed.csv",
             "data_manual_throughput_train_new_vid_processed.csv"]
# filename = "data_manual_throughput_train_new_UAV_0_processed.csv"
def merge_df(filename):
    print(filename)
    clow = "/media/research-student/KingstonSSD/FANET_Dataset/DJISpark_Measured_Throughput_100000_Samples/clow"
    wlau = "/media/research-student/KingstonSSD/FANET_Dataset/DJISpark_Measured_Throughput_100000_Samples/wlau"
    rlim = "/media/research-student/KingstonSSD/FANET_Dataset/DJISpark_Measured_Throughput_100000_Samples/rlim"

    clow_df = pd.read_csv(os.path.join(clow, filename))
    wlau_df = pd.read_csv(os.path.join(wlau, filename))
    rlim_df = pd.read_csv(os.path.join(rlim, filename))

    save_path = "/media/research-student/KingstonSSD/FANET_Dataset/DJISpark_Measured_Throughput_100000_Samples"
    df = pd.concat([wlau_df, rlim_df, clow_df])
    df.to_csv(os.path.join(save_path, filename))

with Pool(16) as pool:
    pool.starmap(merge_df, zip(filenames))
```
