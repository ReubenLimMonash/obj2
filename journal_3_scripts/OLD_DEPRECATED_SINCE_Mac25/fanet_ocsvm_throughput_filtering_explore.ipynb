## Plot single case

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

LINK = "Video" # Downlink / Uplink / Video
USI = 20 # 10 / 20 / 66.7 / 100
MCS = 5 # {1-7}
DATASET_PATH = "/media/research-student/KingstonSSD/ocsvm_models"

x_inlier_npy = np.load("{}/X_inlier_{}_USI-{}_MCS-{}.npy".format(DATASET_PATH, LINK, USI, MCS))
x_outlier_npy = np.load("{}/X_outlier_{}_USI-{}_MCS-{}.npy".format(DATASET_PATH, LINK, USI, MCS))

ax = plt.figure(figsize=(24, 12)).add_subplot(projection='3d')
ax.scatter(x_inlier_npy[:, 0], x_inlier_npy[:, 1], x_inlier_npy[:, 2], marker='o', c='blue', s=2, alpha=0.5)
ax.scatter(x_outlier_npy[:, 0], x_outlier_npy[:, 1], x_outlier_npy[:, 2], marker='s',  c='tab:orange', s=5, alpha=0.8)
```

## Plot all cases

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

DATASET_PATH = "/media/research-student/KingstonSSD/ocsvm_models"
LINK = ["Downlink", "Uplink", "Video"]
USI = [10, 20, 66.7, 100] # 10 / 20 / 66.7 / 100
MCS = np.arange(1,8) # {1-7}

for link in LINK:
    for usi in USI:
        for mcs in MCS:
            inlier_npy = "{}/X_inlier_{}_USI-{}_MCS-{}.npy".format(DATASET_PATH, link, usi, mcs)
            outlier_npy = "{}/X_outlier_{}_USI-{}_MCS-{}.npy".format(DATASET_PATH, link, usi, mcs)
            if os.path.exists(inlier_npy) and os.path.exists(outlier_npy):
                x_inlier_npy = np.load(inlier_npy)
                x_outlier_npy = np.load(outlier_npy)
                ax = plt.figure(figsize=(24, 12)).add_subplot(projection='3d')
                ax.scatter(x_inlier_npy[:, 0], x_inlier_npy[:, 1], x_inlier_npy[:, 2], marker='o', c='blue', s=2, alpha=0.5)
                ax.scatter(x_outlier_npy[:, 0], x_outlier_npy[:, 1], x_outlier_npy[:, 2], marker='s', c='tab:orange', s=2, alpha=0.8)
                plt.title("{}_USI-{}_MCS-{}".format(link, usi, mcs))
```
