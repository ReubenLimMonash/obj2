## Waypoint Mode (No rebroadcasting)

```python
import pandas as pd
import numpy as np

dl_sim_df = pd.read_csv("/media/research-student/One Touch/FANET_Dataset/Dataset_NP10000_DJISpark/data_processed/DJI_Spark_Waypoint_Downlink_Reliability.csv")
ul_sim_df = pd.read_csv("/media/research-student/One Touch/FANET_Dataset/Dataset_NP10000_DJISpark/data_processed/DJI_Spark_Waypoint_Uplink_Reliability.csv")

# Get the reliabilities for DL, UL and Video, and record it in dl_sim_df
dl_sim_df["Reliability_DL"] = dl_sim_df["Num_Reliable"] / dl_sim_df["Num_Sent"]
dl_sim_df["Reliability_UL"] = ul_sim_df["Num_Reliable"] / ul_sim_df["Num_Sent"]

heights = [60, 90, 120, 150, 180, 210, 240, 270, 300]
bitrate_list = [6.5, 13, 19.5, 26, 39, 52, 58.5, 65]
horizontal_dist = np.linspace(0, 700, 71, endpoint=True)
reliability_th = 0.99
results_list = []
for height in heights:
    for bitrate in bitrate_list:
        # ''' Get the max hdist in dl_sim_df, ul_sim_df that fulfills reliability_th '''
        # reliable_df = dl_sim_df.loc[(dl_sim_df["Height"]==height) & (dl_sim_df["Bitrate"]==bitrate) & 
        #                             (dl_sim_df["Reliability_DL"]>=reliability_th) & (dl_sim_df["Reliability_UL"]>=reliability_th)]
        # max_h_dist = reliable_df["Horizontal_Distance"].max()
        # results_list.append({"Height": height, "Bitrate": bitrate, "Max_Horizontal_Distance": max_h_dist})
        ''' Get d_max '''
        d_max = 0
        df = dl_sim_df.loc[(dl_sim_df["Height"]==height) & (dl_sim_df["Bitrate"]==bitrate)]
        for hdist in horizontal_dist:
            df_hdist = df.loc[df["Horizontal_Distance"]==hdist]
            if (df_hdist["Reliability_DL"].values[0] >= reliability_th) & (df_hdist["Reliability_UL"].values[0] >= reliability_th):
                d_max = hdist
            else:
                break

        results_list.append({"Height": height, "Bitrate": bitrate, "Max_Horizontal_Distance": d_max})

results_df = pd.DataFrame(results_list)
results_df.to_csv("/home/research-student/omnet-fanet/data-processing-scripts/waypoint_max_hdist/Waypoint_Max_HDist_fm_Sim_99.csv")

```

## Manual Control Mode No Retransmission

```python
import pandas as pd
import numpy as np

dl_sim_df = pd.read_csv("/media/research-student/KingstonSSD/FANET_Dataset/Dataset_NP100000_Manual_Control_No_Retransmission/data_processed/Downlink_Reliability.csv")
ul_sim_df = pd.read_csv("/media/research-student/KingstonSSD/FANET_Dataset/Dataset_NP100000_Manual_Control_No_Retransmission/data_processed/Uplink_Reliability.csv")
vid_sim_df = pd.read_csv("/media/research-student/KingstonSSD/FANET_Dataset/Dataset_NP100000_Manual_Control_No_Retransmission/data_processed/Video_Reliability.csv")

# Get the reliabilities for DL, UL and Video, and record it in dl_sim_df
dl_sim_df["Reliability_DL"] = dl_sim_df["Num_Reliable"] / dl_sim_df["Num_Sent"]
dl_sim_df["Reliability_UL"] = ul_sim_df["Num_Reliable"] / ul_sim_df["Num_Sent"]
dl_sim_df["Reliability_VID"] = vid_sim_df["Num_Reliable"] / vid_sim_df["Num_Sent"]

heights = [60, 90, 120, 150, 180, 210, 240, 270, 300]
bitrate_list = [6.5, 13, 19.5, 26, 39, 52, 58.5, 65]
reliability_th = 0.50
results_list = []
for height in heights:
    for bitrate in bitrate_list:
        # Get the max hdist in dl_sim_df, ul_sim_df and vid_sim_df that fulfills reliability_th
        reliable_df = dl_sim_df.loc[(dl_sim_df["Height"]==height) & (dl_sim_df["Bitrate"]==bitrate) & (dl_sim_df["UAV_Sending_Interval"]==100) & 
                                    (dl_sim_df["Reliability_DL"]>=reliability_th) & (dl_sim_df["Reliability_UL"]>=reliability_th) & 
                                    (dl_sim_df["Reliability_VID"]>=reliability_th)]
        max_h_dist = reliable_df["Horizontal_Distance"].max()
        results_list.append({"Height": height, "Bitrate": bitrate, "Max_Horizontal_Distance": max_h_dist})

results_df = pd.DataFrame(results_list)
results_df.to_csv("Manual_Control_No_Retransmission_Max_HDist_fm_Sim_50.csv")
```
