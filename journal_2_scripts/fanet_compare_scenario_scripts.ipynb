### Compare the MCS Actions from Two Folders, Checking If They Are The Same

```python
import xml.etree.ElementTree as ET
import glob
import os
import numpy as np

xml_folder_1 = "/home/research-student/omnet-fanet/scenario_scripts_mcs_adaptation/scenario_scripts_1m_calibrated_4_specificity_999_v2"
xml_folder_2 = "/home/research-student/omnet-fanet/scenario_scripts_mcs_adaptation/scenario_scripts_1m_calibrated_4_specificity_999"

xml_files = [file.split("/")[-1] for file in glob.glob(os.path.join(xml_folder_1, "*.xml"))]
comparison = []
for file in xml_files:
    xml_1 = ET.tostring(ET.parse(os.path.join(xml_folder_1, file)).getroot(), 'utf-8')
    xml_2 = ET.tostring(ET.parse(os.path.join(xml_folder_2, file)).getroot(), 'utf-8')
    comparison.append(xml_1 == xml_2)
np.all(comparison)
```
