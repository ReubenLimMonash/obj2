## Get max hdist at which reliability reaches below 0.1% regardless of parameters 

```python
import pandas as pd
import numpy as np
import os

DATASET_PATH = "/media/research-student/KingstonSSD/FANET_Dataset/Dataset_NP100000_ParrotAR2/data_processed"
USI = [10.0, 20.0, 66.7, 100.0]
BITRATE = [6.5, 13.0, 19.5, 26.0, 39.0, 52.0, 58.5, 65.0]
HEIGHT = [60, 90, 120, 150, 180, 210, 240, 270, 300]
RELIABILITY_THRESHOLD = 0.001 # Threshold to consider limits of hdist for reliability in Obj 2

dl_df = pd.read_csv(os.path.join(DATASET_PATH, "Downlink_Reliability.csv"))
dl_df["Reliability"] = dl_df["Num_Reliable"] / dl_df["Num_Sent"]
max_hdists = []
for usi in USI:
    for bitrate in BITRATE:
        for height in HEIGHT:
            df = dl_df.loc[(dl_df["Height"] == height) & (dl_df["Bitrate"] == bitrate) & (dl_df["UAV_Sending_Interval"] == usi)]
            df_fail = df.loc[df["Reliability"] < RELIABILITY_THRESHOLD]
            max_hdists.append(df_fail["Horizontal_Distance"].min())

dl_max_hdist = np.max(max_hdists)

ul_df = pd.read_csv(os.path.join(DATASET_PATH, "Uplink_Reliability.csv"))
ul_df["Reliability"] = ul_df["Num_Reliable"] / ul_df["Num_Sent"]
max_hdists = []
for usi in USI:
    for bitrate in BITRATE:
        for height in HEIGHT:
            df = ul_df.loc[(ul_df["Height"] == height) & (ul_df["Bitrate"] == bitrate) & (ul_df["UAV_Sending_Interval"] == usi)]
            df_fail = df.loc[df["Reliability"] < RELIABILITY_THRESHOLD]
            max_hdists.append(df_fail["Horizontal_Distance"].min())

ul_max_hdist = np.max(max_hdists)

vid_df = pd.read_csv(os.path.join(DATASET_PATH, "Video_Reliability.csv"))
vid_df["Reliability"] = vid_df["Num_Reliable"] / vid_df["Num_Sent"]
max_hdists = []
for usi in USI:
    for bitrate in BITRATE:
        for height in HEIGHT:
            df = vid_df.loc[(vid_df["Height"] == height) & (vid_df["Bitrate"] == bitrate) & (vid_df["UAV_Sending_Interval"] == usi)]
            df_fail = df.loc[df["Reliability"] < RELIABILITY_THRESHOLD]
            max_hdists.append(df_fail["Horizontal_Distance"].min())

vid_max_hdist = np.max(max_hdists)

print(dl_max_hdist, ul_max_hdist, vid_max_hdist)
```

```python
import pandas as pd
import numpy as np
import os

DATASET_10E5 = "/media/research-student/KingstonSSD/FANET_Dataset/Dataset_NP100000_DJISpark/data_processed"
DATASET_10E4 = "/media/research-student/One Touch/FANET_Dataset/Dataset_NP10000_DJISpark/data_processed"

# Downlink
dl_df_e5 = pd.read_csv(DATASET_10E5 + "/Downlink_Reliability.csv")
dl_df_e4 = pd.read_csv(DATASET_10E4 + "/DJI_Spark_Downlink_Reliability.csv")
dl_df_e5 = dl_df_e5.loc[dl_df_e5["Horizontal_Distance"] <= 700]
dl_df_e4 = dl_df_e4.loc[dl_df_e4["Horizontal_Distance"] <= 700]
dl_df_e5["Reliability"] = dl_df_e5["Num_Reliable"] / dl_df_e5["Num_Sent"]
dl_df_e4["Reliability"] = dl_df_e4["Num_Reliable"] / dl_df_e4["Num_Sent"]
dl_df_e5.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
dl_df_e4.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
dl_df_e5["Abs_Err"] = np.abs(dl_df_e5["Reliability"].values - dl_df_e4["Reliability"].values)
dl_mean_abs_err = dl_df_e5["Abs_Err"].mean()
dl_max_abs_err = dl_df_e5["Abs_Err"].max()

# Uplink
ul_df_e5 = pd.read_csv(DATASET_10E5 + "/Uplink_Reliability.csv")
ul_df_e4 = pd.read_csv(DATASET_10E4 + "/DJI_Spark_Uplink_Reliability.csv")
ul_df_e5 = ul_df_e5.loc[ul_df_e5["Horizontal_Distance"] <= 700]
ul_df_e4 = ul_df_e4.loc[ul_df_e4["Horizontal_Distance"] <= 700]
ul_df_e5["Reliability"] = ul_df_e5["Num_Reliable"] / ul_df_e5["Num_Sent"]
ul_df_e4["Reliability"] = ul_df_e4["Num_Reliable"] / ul_df_e4["Num_Sent"]
ul_df_e5.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
ul_df_e4.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
ul_df_e5["Abs_Err"] = np.abs(ul_df_e5["Reliability"].values - ul_df_e4["Reliability"].values)
ul_mean_abs_err = ul_df_e5["Abs_Err"].mean()
ul_max_abs_err = ul_df_e5["Abs_Err"].max()

# Video
vid_df_e5 = pd.read_csv(DATASET_10E5 + "/Video_Reliability.csv")
vid_df_e4 = pd.read_csv(DATASET_10E4 + "/DJI_Spark_Video_Reliability.csv")
vid_df_e5 = vid_df_e5.loc[vid_df_e5["Horizontal_Distance"] <= 700]
vid_df_e4 = vid_df_e4.loc[vid_df_e4["Horizontal_Distance"] <= 700]
vid_df_e5["Reliability"] = vid_df_e5["Num_Reliable"] / vid_df_e5["Num_Sent"]
vid_df_e4["Reliability"] = vid_df_e4["Num_Reliable"] / vid_df_e4["Num_Sent"]
vid_df_e5.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
vid_df_e4.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
vid_df_e5["Abs_Err"] = np.abs(vid_df_e5["Reliability"].values - vid_df_e4["Reliability"].values)
vid_mean_abs_err = vid_df_e5["Abs_Err"].mean()
vid_max_abs_err = vid_df_e5["Abs_Err"].max()

print(dl_mean_abs_err, dl_max_abs_err)
print(ul_mean_abs_err, ul_max_abs_err)
print(vid_mean_abs_err, vid_max_abs_err)

djispark_meanAE = [dl_mean_abs_err, ul_mean_abs_err, vid_mean_abs_err]
```

```python
import pandas as pd
import numpy as np
import os

DATASET_10E5 = "/media/research-student/KingstonSSD/FANET_Dataset/Dataset_NP100000_DJIMavicAir/test_dataset_{}_processed"
DATASET_10E4 = "/media/research-student/One Touch/FANET_Dataset/Dataset_NP10000_DJIMavicAir/test_dataset_{}_processed"

# Downlink
dl_df_e5_1 = pd.read_csv(DATASET_10E5.format(1) + "/Downlink_Reliability.csv")
dl_df_e5_2 = pd.read_csv(DATASET_10E5.format(2) + "/Downlink_Reliability.csv")
dl_df_e4_1 = pd.read_csv(DATASET_10E4.format(1) + "/Downlink_Reliability.csv")
dl_df_e4_2 = pd.read_csv(DATASET_10E4.format(2) + "/Downlink_Reliability.csv")
dl_df_e5 = pd.concat([dl_df_e5_1, dl_df_e5_2])
dl_df_e4 = pd.concat([dl_df_e4_1, dl_df_e4_2])
dl_df_e5 = dl_df_e5.loc[dl_df_e5["Horizontal_Distance"] <= 700]
dl_df_e4 = dl_df_e4.loc[dl_df_e4["Horizontal_Distance"] <= 700]
dl_df_e5["Reliability"] = dl_df_e5["Num_Reliable"] / dl_df_e5["Num_Sent"]
dl_df_e4["Reliability"] = dl_df_e4["Num_Reliable"] / dl_df_e4["Num_Sent"]
dl_df_e5.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
dl_df_e4.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
dl_df_e5["Abs_Err"] = np.abs(dl_df_e5["Reliability"].values - dl_df_e4["Reliability"].values)
dl_mean_abs_err = dl_df_e5["Abs_Err"].mean()
dl_max_abs_err = dl_df_e5["Abs_Err"].max()

# Uplink
ul_df_e5_1 = pd.read_csv(DATASET_10E5.format(1) + "/Uplink_Reliability.csv")
ul_df_e5_2 = pd.read_csv(DATASET_10E5.format(2) + "/Uplink_Reliability.csv")
ul_df_e4_1 = pd.read_csv(DATASET_10E4.format(1) + "/Uplink_Reliability.csv")
ul_df_e4_2 = pd.read_csv(DATASET_10E4.format(2) + "/Uplink_Reliability.csv")
ul_df_e5 = pd.concat([ul_df_e5_1, ul_df_e5_2])
ul_df_e4 = pd.concat([ul_df_e4_1, ul_df_e4_2])
ul_df_e5 = ul_df_e5.loc[ul_df_e5["Horizontal_Distance"] <= 700]
ul_df_e4 = ul_df_e4.loc[ul_df_e4["Horizontal_Distance"] <= 700]
ul_df_e5["Reliability"] = ul_df_e5["Num_Reliable"] / ul_df_e5["Num_Sent"]
ul_df_e4["Reliability"] = ul_df_e4["Num_Reliable"] / ul_df_e4["Num_Sent"]
ul_df_e5.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
ul_df_e4.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
ul_df_e5["Abs_Err"] = np.abs(ul_df_e5["Reliability"].values - ul_df_e4["Reliability"].values)
ul_mean_abs_err = ul_df_e5["Abs_Err"].mean()
ul_max_abs_err = ul_df_e5["Abs_Err"].max()

# Video
vid_df_e5_1 = pd.read_csv(DATASET_10E5.format(1) + "/Video_Reliability.csv")
vid_df_e5_2 = pd.read_csv(DATASET_10E5.format(2) + "/Video_Reliability.csv")
vid_df_e4_1 = pd.read_csv(DATASET_10E4.format(1) + "/Video_Reliability.csv")
vid_df_e4_2 = pd.read_csv(DATASET_10E4.format(2) + "/Video_Reliability.csv")
vid_df_e5 = pd.concat([vid_df_e5_1, vid_df_e5_2])
vid_df_e4 = pd.concat([vid_df_e4_1, vid_df_e4_2])
vid_df_e5 = vid_df_e5.loc[vid_df_e5["Horizontal_Distance"] <= 700]
vid_df_e4 = vid_df_e4.loc[vid_df_e4["Horizontal_Distance"] <= 700]
vid_df_e5["Reliability"] = vid_df_e5["Num_Reliable"] / vid_df_e5["Num_Sent"]
vid_df_e4["Reliability"] = vid_df_e4["Num_Reliable"] / vid_df_e4["Num_Sent"]
vid_df_e5.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
vid_df_e4.sort_values(by=["UAV_Sending_Interval", "Bitrate", "Height"], inplace=True)
vid_df_e5["Abs_Err"] = np.abs(vid_df_e5["Reliability"].values - vid_df_e4["Reliability"].values)
vid_mean_abs_err = vid_df_e5["Abs_Err"].mean()
vid_max_abs_err = vid_df_e5["Abs_Err"].max()

print(dl_mean_abs_err, dl_max_abs_err)
print(ul_mean_abs_err, ul_max_abs_err)
print(vid_mean_abs_err, vid_max_abs_err)

djimavicair_meanAE = [dl_mean_abs_err, ul_mean_abs_err, vid_mean_abs_err]
```
